````python
%%sql

-- ============================================================
-- STEP 1: parent edges
-- ============================================================
DROP TABLE IF EXISTS tmp_parent_edges;
CREATE TEMP TABLE tmp_parent_edges AS
SELECT
    r.person_id,
    r.related_person_id AS parent_id
FROM core.relationship r
JOIN core.mapkinship mk ON r.relationship = mk.relationship
WHERE mk.kinship = 'PARENT';

CREATE INDEX idx_tmp_parent_person   ON tmp_parent_edges (person_id);
CREATE INDEX idx_tmp_parent_parentid ON tmp_parent_edges (parent_id);

-- ============================================================
-- STEP 2: ancestor chain (max depth 10)
-- ============================================================
DROP TABLE IF EXISTS tmp_ancestor_chain;
CREATE TEMP TABLE tmp_ancestor_chain AS
WITH RECURSIVE ancestor_chain AS (
    SELECT person_id, parent_id AS ancestor_id, 1 AS depth
    FROM tmp_parent_edges

    UNION ALL

    SELECT ac.person_id, pe.parent_id AS ancestor_id, ac.depth + 1
    FROM ancestor_chain ac
    JOIN tmp_parent_edges pe ON ac.ancestor_id = pe.person_id
    WHERE ac.depth < 10
)
SELECT * FROM ancestor_chain;

CREATE INDEX idx_tmp_anc_person   ON tmp_ancestor_chain (person_id);
CREATE INDEX idx_tmp_anc_ancestor ON tmp_ancestor_chain (ancestor_id);
CREATE INDEX idx_tmp_anc_depth    ON tmp_ancestor_chain (depth);

-- ============================================================
-- STEP 3: sibling edges
-- ============================================================
DROP TABLE IF EXISTS tmp_sibling_edges;
CREATE TEMP TABLE tmp_sibling_edges AS
SELECT DISTINCT
    p1.person_id,
    p2.person_id AS sibling_id
FROM tmp_parent_edges p1
JOIN tmp_parent_edges p2
    ON p1.parent_id  = p2.parent_id
    AND p1.person_id <> p2.person_id;

CREATE INDEX idx_tmp_sib_person  ON tmp_sibling_edges (person_id);
CREATE INDEX idx_tmp_sib_sibling ON tmp_sibling_edges (sibling_id);

-- ============================================================
-- STEP 3b: spouse edges
-- ============================================================
DROP TABLE IF EXISTS tmp_spouse_edges;
CREATE TEMP TABLE tmp_spouse_edges AS
SELECT
    r.person_id,
    r.related_person_id AS spouse_id
FROM core.relationship r
JOIN core.mapkinship mk
    ON r.relationship = mk.relationship
WHERE mk.kinship = 'SPOUSE';

CREATE INDEX idx_tmp_spouse_person ON tmp_spouse_edges (person_id);
CREATE INDEX idx_tmp_spouse_spouse ON tmp_spouse_edges (spouse_id);

-- ============================================================
-- STEP 4: same generation chain
-- ============================================================
DROP TABLE IF EXISTS tmp_samegen_chain;
CREATE TEMP TABLE tmp_samegen_chain AS
WITH RECURSIVE samegen_chain AS (
    SELECT person_id, sibling_id AS samegen_id, 1 AS depth
    FROM tmp_sibling_edges

    UNION ALL

    SELECT
        p1.person_id,
        p2.person_id AS samegen_id,
        sc.depth + 1
    FROM samegen_chain sc
    JOIN tmp_parent_edges p1 ON p1.parent_id = sc.person_id
    JOIN tmp_parent_edges p2 ON p2.parent_id = sc.samegen_id
    WHERE p1.person_id <> p2.person_id
      AND sc.depth < 3
)
SELECT DISTINCT person_id, samegen_id FROM samegen_chain;

CREATE INDEX idx_tmp_sg_person  ON tmp_samegen_chain (person_id);
CREATE INDEX idx_tmp_sg_samegen ON tmp_samegen_chain (samegen_id);

-- ============================================================
-- STEP 4b: aunt/uncle edges
-- ============================================================
DROP TABLE IF EXISTS tmp_aunt_uncle_edges;
CREATE TEMP TABLE tmp_aunt_uncle_edges AS
SELECT DISTINCT
    child.person_id,
    sib.sibling_id AS aunt_uncle_id
FROM tmp_parent_edges child
JOIN tmp_sibling_edges sib
    ON sib.person_id = child.parent_id
WHERE child.person_id <> sib.sibling_id;

CREATE INDEX idx_tmp_au_person    ON tmp_aunt_uncle_edges (person_id);
CREATE INDEX idx_tmp_au_auntuncle ON tmp_aunt_uncle_edges (aunt_uncle_id);

-- ============================================================
-- STEP 5: base kinships
-- ============================================================
DROP TABLE IF EXISTS tmp_base_kinships;
CREATE TEMP TABLE tmp_base_kinships AS

-- direct mapped kinships
SELECT r.person_id, mk.kinship, r.related_person_id
FROM core.relationship r
JOIN core.mapkinship mk ON r.relationship = mk.relationship
WHERE mk.kinship IN (
    'PARENT','CHILD',
    'GRANDPARENT','GRANDCHILD',
    'ANCESTOR','DESCENDANT',
    'SIBLING',
    'COUSIN',
    'AUNT_UNCLE','NIECE_NEPHEW',
    'SPOUSE',
    'PARENT_INLAW','CHILD_INLAW','SIBLING_INLAW','RELATIVE_INLAW',
    'RELATIVE'
)

UNION ALL

-- inferred parent
SELECT person_id, 'PARENT', ancestor_id
FROM tmp_ancestor_chain WHERE depth = 1

UNION ALL

-- inferred grandparent
SELECT person_id, 'GRANDPARENT', ancestor_id
FROM tmp_ancestor_chain WHERE depth = 2

UNION ALL

-- inferred ancestor
SELECT person_id, 'ANCESTOR', ancestor_id
FROM tmp_ancestor_chain WHERE depth >= 3

UNION ALL

-- inferred child
SELECT parent_id, 'CHILD', person_id
FROM tmp_parent_edges

UNION ALL

-- inferred grandchild
SELECT ancestor_id, 'GRANDCHILD', person_id
FROM tmp_ancestor_chain WHERE depth = 2

UNION ALL

-- inferred descendant
SELECT ancestor_id, 'DESCENDANT', person_id
FROM tmp_ancestor_chain WHERE depth >= 3

UNION ALL

-- siblings
SELECT person_id, 'SIBLING', sibling_id
FROM tmp_sibling_edges

UNION ALL

-- cousins
SELECT DISTINCT sc.person_id, 'COUSIN', sc.samegen_id
FROM tmp_samegen_chain sc
WHERE NOT EXISTS (
    SELECT 1 FROM tmp_sibling_edges se
    WHERE se.person_id = sc.person_id
      AND se.sibling_id = sc.samegen_id
)
AND NOT EXISTS (
    SELECT 1 FROM tmp_aunt_uncle_edges au
    WHERE au.person_id = sc.person_id
      AND au.aunt_uncle_id = sc.samegen_id
)
AND NOT EXISTS (
    SELECT 1 FROM tmp_aunt_uncle_edges au
    WHERE au.person_id = sc.samegen_id
      AND au.aunt_uncle_id = sc.person_id
)

UNION ALL

-- aunt uncle
SELECT person_id, 'AUNT_UNCLE', aunt_uncle_id
FROM tmp_aunt_uncle_edges

UNION ALL

-- niece nephew
SELECT aunt_uncle_id, 'NIECE_NEPHEW', person_id
FROM tmp_aunt_uncle_edges

UNION ALL

-- parent in law
SELECT
    s.person_id,
    'PARENT_INLAW',
    pe.parent_id
FROM tmp_spouse_edges s
JOIN tmp_parent_edges pe
ON pe.person_id = s.spouse_id

UNION ALL

-- child in law
SELECT
    s.person_id,
    'CHILD_INLAW',
    pe.person_id
FROM tmp_spouse_edges s
JOIN tmp_parent_edges pe
ON pe.parent_id = s.spouse_id

UNION ALL

-- sibling in law
SELECT
    s.person_id,
    'SIBLING_INLAW',
    se.sibling_id
FROM tmp_spouse_edges s
JOIN tmp_sibling_edges se
ON se.person_id = s.spouse_id;

CREATE INDEX idx_tmp_bk_person  ON tmp_base_kinships (person_id);
CREATE INDEX idx_tmp_bk_related ON tmp_base_kinships (related_person_id);
CREATE INDEX idx_tmp_bk_kinship ON tmp_base_kinships (kinship);

-- ============================================================
-- STEP 6: propagate down
-- ============================================================
DROP TABLE IF EXISTS tmp_propagated;
CREATE TEMP TABLE tmp_propagated AS
SELECT DISTINCT
    ac.person_id,
    CASE
        WHEN bk.kinship = 'PARENT'         THEN 'GRANDPARENT'
        WHEN bk.kinship = 'GRANDPARENT'    THEN 'ANCESTOR'
        WHEN bk.kinship = 'ANCESTOR'       THEN 'ANCESTOR'
        WHEN bk.kinship = 'SIBLING'        THEN 'COUSIN'
        WHEN bk.kinship = 'COUSIN'         THEN 'COUSIN'
        WHEN bk.kinship = 'AUNT_UNCLE'     THEN 'AUNT_UNCLE'
        WHEN bk.kinship = 'NIECE_NEPHEW'   THEN 'RELATIVE'
        WHEN bk.kinship = 'RELATIVE'       THEN 'RELATIVE'
        ELSE 'RELATIVE'
    END AS kinship,
    bk.related_person_id
FROM tmp_ancestor_chain ac
JOIN tmp_base_kinships bk ON bk.person_id = ac.ancestor_id
WHERE bk.kinship NOT IN (
    'CHILD','GRANDCHILD','DESCENDANT',
    'CHILD_INLAW','PARENT_INLAW','SIBLING_INLAW','RELATIVE_INLAW'
)
AND ac.person_id <> bk.related_person_id;

-- ============================================================
-- STEP 7: final insert
-- ============================================================
DROP TABLE IF EXISTS core.kinship CASCADE;

CREATE TABLE core.kinship (
    person_id VARCHAR(50),
    kinship VARCHAR(20),
    related_person_id VARCHAR(50),
    PRIMARY KEY (person_id, related_person_id)
);

INSERT INTO core.kinship
SELECT person_id, kinship, related_person_id
FROM (
    SELECT
        person_id,
        kinship,
        related_person_id,
        ROW_NUMBER() OVER (
            PARTITION BY person_id, related_person_id
            ORDER BY CASE kinship
                WHEN 'SPOUSE'         THEN 1
                WHEN 'PARENT'         THEN 2
                WHEN 'CHILD'          THEN 3
                WHEN 'SIBLING'        THEN 4
                WHEN 'GRANDPARENT'    THEN 5
                WHEN 'GRANDCHILD'     THEN 6
                WHEN 'AUNT_UNCLE'     THEN 7
                WHEN 'NIECE_NEPHEW'   THEN 8
                WHEN 'COUSIN'         THEN 9
                WHEN 'PARENT_INLAW'   THEN 10
                WHEN 'CHILD_INLAW'    THEN 11
                WHEN 'SIBLING_INLAW'  THEN 12
                WHEN 'RELATIVE_INLAW' THEN 13
                WHEN 'ANCESTOR'       THEN 14
                WHEN 'DESCENDANT'     THEN 15
                WHEN 'RELATIVE'       THEN 16
                ELSE 99
            END
        ) AS rn
    FROM (
        SELECT * FROM tmp_base_kinships
        UNION ALL
        SELECT * FROM tmp_propagated
    ) combined
    WHERE person_id <> related_person_id
) ranked
WHERE rn = 1;

-- ============================================================
-- STEP 8: cleanup
-- ============================================================
DROP TABLE IF EXISTS tmp_parent_edges;
DROP TABLE IF EXISTS tmp_ancestor_chain;
DROP TABLE IF EXISTS tmp_sibling_edges;
DROP TABLE IF EXISTS tmp_samegen_chain;
DROP TABLE IF EXISTS tmp_aunt_uncle_edges;
DROP TABLE IF EXISTS tmp_spouse_edges;
DROP TABLE IF EXISTS tmp_base_kinships;
DROP TABLE IF EXISTS tmp_propagated;
````

## DDL nepo baby score

````python
%%sql
CREATE TABLE IF NOT EXISTS core.neposcore AS
SELECT
    k.person_id,

    ROUND(SUM(
        COALESCE(pp.person_point,0) *
        CASE k.kinship
            WHEN 'PARENT'        THEN 1.0
            WHEN 'CHILD'         THEN 1.0
            WHEN 'SIBLING'       THEN 0.5
            WHEN 'GRANDPARENT'   THEN 0.5
            WHEN 'GRANDCHILD'    THEN 0.5
            WHEN 'AUNT_UNCLE'    THEN 0.25
            WHEN 'NIECE_NEPHEW'  THEN 0.25
            WHEN 'COUSIN'        THEN 0.25
            WHEN 'ANCESTOR'      THEN 0.25
            WHEN 'DESCENDANT'    THEN 0.25
            WHEN 'RELATIVE'      THEN 0.1



            ELSE 0
        END
    )) AS nepo_score,

    ROUND(SUM(
        CASE WHEN k.kinship = 'PARENT'
        THEN COALESCE(pp.person_point,0) ELSE 0 END
    )) AS parent_score,

    ROUND(SUM(
        CASE WHEN k.kinship = 'GRANDPARENT'
        THEN COALESCE(pp.person_point,0) * 0.5 ELSE 0 END
    )) AS grandparent_score,

    ROUND(SUM(
        CASE WHEN k.kinship = 'ANCESTOR'
        THEN COALESCE(pp.person_point,0) * 0.25 ELSE 0 END
    )) AS ancestor_score,

    ROUND(SUM(
        CASE WHEN k.kinship = 'RELATIVE'
        THEN COALESCE(pp.person_point,0) * 0.1 ELSE 0 END
    )) AS relative_score

FROM core.kinship k
LEFT JOIN core.personpoint pp
       ON pp.person_id = k.related_person_id
GROUP BY k.person_id;
````

### breakdown points

````python
%%sql
DROP TABLE IF EXISTS core.personmoviepoint CASCADE;

CREATE TABLE IF NOT EXISTS core.personmoviepoint (
    person_id          VARCHAR(20),
    tconst             VARCHAR(10),
    primary_title      VARCHAR(255),
    ordering           INT,
    ordering_weight    NUMERIC(10,0),
    movie_point        NUMERIC(10,0),
    weighted_point     NUMERIC(10,0),
    PRIMARY KEY (person_id, tconst)
);

INSERT INTO core.personmoviepoint (
    person_id,
    tconst,
    primary_title,
    ordering,
    ordering_weight,
    movie_point,
    weighted_point
)
WITH
moviepoints AS (
    SELECT
        t.tconst,
        t.primary_title,
        (
            COALESCE(LOG(r.num_votes::NUMERIC) * r.average_rating, 0) +
            COALESCE(SUM(CASE WHEN m.type = 'Award' THEN 10 ELSE 3 END), 0)
        ) AS movie_point
    FROM core.title t
    LEFT JOIN core.rating r     ON t.tconst = r.tconst
    LEFT JOIN core.movieaward m ON t.tconst = m.tconst
    GROUP BY t.tconst, t.primary_title, r.num_votes, r.average_rating
),
cast_weights AS (
    SELECT
        tconst,
        ordering,
        (COUNT(*) OVER (PARTITION BY tconst) - ordering + 1.0) /
        (COUNT(*) OVER (PARTITION BY tconst) *
            (COUNT(*) OVER (PARTITION BY tconst) + 1.0) / 2.0) AS weight
    FROM core.principal
),
person_movie_raw AS (
    SELECT
        p.person_id,
        mp.tconst,
        mp.primary_title,
        pr.ordering,
        cw.weight                        AS ordering_weight,
        mp.movie_point                   AS movie_point,
        cw.weight * mp.movie_point       AS weighted_point,
        ROW_NUMBER() OVER (
            PARTITION BY p.person_id, mp.tconst
            ORDER BY pr.ordering ASC
        ) AS rn
    FROM core.person p
    JOIN core.principal pr  ON p.nconst   = pr.nconst
    JOIN moviepoints mp     ON pr.tconst  = mp.tconst
    JOIN cast_weights cw    ON cw.tconst  = pr.tconst
                           AND cw.ordering = pr.ordering
)
SELECT
    person_id,
    tconst,
    primary_title,
    ordering,
    ROUND(ordering_weight, 2)  AS ordering_weight,
    ROUND(movie_point, 2)      AS movie_point,
    ROUND(weighted_point, 2)   AS weighted_point
FROM person_movie_raw
WHERE rn = 1
ORDER BY person_id, weighted_point DESC;
````

## family


````python
%%sql
/* ============================================================
  SECTION 6 — FAMILY DYNASTIES (DERIVED FROM LAST NAME, IGNORING SUFFIXES)
  ------------------------------------------------------------
  Depends On: core.person, core.neposcore
  Purpose: Identify prominent families using last name, handling suffixes
  ============================================================ */


WITH person_family AS (
   SELECT
       p.person_id,
       p.name,
       ns.nepo_score,
       -- Split name into array
       string_to_array(p.name, ' ') AS name_parts
   FROM core.person p
            LEFT JOIN core.neposcore ns
                      ON ns.person_id = p.person_id
   WHERE ns.nepo_score IS NOT NULL
),
    family_extracted AS (
        SELECT
            person_id,
            name,
            nepo_score,
            CASE
                WHEN upper(name_parts[array_length(name_parts,1)]) IN ('JR.','SR.','II','III','IV','V')
                    THEN name_parts[array_length(name_parts,1)-1]
                ELSE name_parts[array_length(name_parts,1)]
                END AS family_name
        FROM person_family
    )
SELECT
   family_name,
   COUNT(DISTINCT person_id) AS family_members,
   AVG(nepo_score)::numeric(10,2) AS avg_family_nepo_score,
   MAX(nepo_score) ::numeric(10,2) AS top_family_nepo_score
FROM family_extracted
WHERE family_name IS NOT NULL
GROUP BY family_name
HAVING COUNT(DISTINCT person_id) >= 2
ORDER BY avg_family_nepo_score DESC NULLS LAST,
        family_members DESC
LIMIT 100;
````

````python
%%sql
WITH family_groups AS (
    -- Find all connected family components via core.kinship
    -- Each person + their relatives form a "family"
    SELECT DISTINCT
        LEAST(k.person_id, k.related_person_id)    AS person_a,
        GREATEST(k.person_id, k.related_person_id) AS person_b
    FROM core.kinship k
    WHERE k.kinship IN ('PARENT', 'GRANDPARENT', 'ANCESTOR', 'SAMEGEN', 'RELATIVE')
),
-- Assign a family_id = the smallest person_id in the connected group
-- Use recursive CTE to find all connected components
family_components AS (
    SELECT
        person_a AS person_id,
        person_a AS family_id
    FROM family_groups

    UNION

    SELECT
        person_b AS person_id,
        person_a AS family_id
    FROM family_groups
),
-- For each person, pick the smallest family_id as their family root
family_assigned AS (
    SELECT
        person_id,
        MIN(family_id) AS family_id
    FROM family_components
    GROUP BY person_id
),
-- Get the name of the family root person as the family name
family_named AS (
    SELECT
        fa.person_id,
        fa.family_id,
        p_root.name AS family_name,
        p.name      AS person_name,
        ns.nepo_score
    FROM family_assigned fa
    JOIN core.person p      ON p.person_id      = fa.person_id
    JOIN core.person p_root ON p_root.person_id = fa.family_id
    LEFT JOIN core.neposcore ns ON ns.person_id = fa.person_id
    WHERE ns.nepo_score IS NOT NULL
)
SELECT
    family_id,
    family_name                             AS family_root_person,
    COUNT(DISTINCT person_id)               AS family_members,
    AVG(nepo_score)::NUMERIC(10,2)          AS avg_family_nepo_score,
    MAX(nepo_score)      ::NUMERIC(10,2)                    AS top_family_nepo_score,
    STRING_AGG(person_name, ', '
        ORDER BY nepo_score DESC NULLS LAST) AS members
FROM family_named
GROUP BY family_id, family_name
HAVING COUNT(DISTINCT person_id) >= 2
ORDER BY avg_family_nepo_score DESC NULLS LAST,
         family_members DESC
LIMIT 100;
````

## FUNCTION


### nepo breakdown

````python
def nb(identifier, blur=False):
    """
    Get nepo score breakdown for a person.

    Usage:
        nb('Q116869845')              # by Wikidata ID
        nb('Blake Lively')            # by exact name
        nb('blake', blur=True)        # partial name match
    """

    # Resolve name to person_id if not a QID
    if not str(identifier).startswith('Q'):
        if blur:
            resolve_sql = text("""
                SELECT person_id, name FROM core.person
                WHERE name ILIKE :name
                ORDER BY name
                LIMIT 10
            """)
            search_term = f'%{identifier}%'
        else:
            resolve_sql = text("""
                SELECT person_id, name FROM core.person
                WHERE name ILIKE :name
                LIMIT 1
            """)
            search_term = identifier

        with engine.connect() as conn:
            results = conn.execute(resolve_sql, {"name": search_term}).fetchall()

            if not results:
                print(f"No person found with name: {identifier}")
                return None

            if blur and len(results) > 1:
                print("Multiple matches found:")
                for i, row in enumerate(results):
                    print(f"  [{i}] {row[0]} — {row[1]}")
                choice = int(input("Enter number to select: "))
                person_id = results[choice][0]
                print(f"Selected: {results[choice][1]} ({person_id})")
            else:
                person_id = results[0][0]
                print(f"Resolved '{identifier}' → {person_id} ({results[0][1]})")
    else:
        person_id = identifier

    # ---------------------------------------------------------
    # Updated query — all INLAW weights set to 0
    # ---------------------------------------------------------
    query = text("""
        SELECT
            k.person_id,
            p1.name AS person_name,
            k.kinship,
            k.related_person_id,
            p2.name AS related_person_name,
            ROUND(COALESCE(pp.person_point, 0)) AS person_point,

            -- Weight model
            CASE
                WHEN k.kinship = 'PARENT'         THEN 1.0
                WHEN k.kinship = 'CHILD'          THEN 1.0
                WHEN k.kinship = 'SIBLING'        THEN 0.5
                WHEN k.kinship = 'GRANDPARENT'    THEN 0.5
                WHEN k.kinship = 'GRANDCHILD'     THEN 0.5
                WHEN k.kinship = 'AUNT_UNCLE'     THEN 0.25
                WHEN k.kinship = 'NIECE_NEPHEW'   THEN 0.25
                WHEN k.kinship = 'COUSIN'         THEN 0.25
                WHEN k.kinship = 'ANCESTOR'       THEN 0.25
                WHEN k.kinship = 'DESCENDANT'     THEN 0.25
                WHEN k.kinship = 'RELATIVE'       THEN 0.1
                WHEN k.kinship = 'SPOUSE'         THEN 0.5
                -- INLAW now zero
                WHEN k.kinship IN ('PARENT_INLAW', 'CHILD_INLAW', 'SIBLING_INLAW', 'RELATIVE_INLAW') THEN 0
                ELSE 0
            END AS kinship_weight,

            -- Contribution
            ROUND(COALESCE(pp.person_point,0) *
                CASE
                    WHEN k.kinship = 'PARENT'         THEN 1.0
                    WHEN k.kinship = 'CHILD'          THEN 1.0
                    WHEN k.kinship = 'SIBLING'        THEN 0.5
                    WHEN k.kinship = 'GRANDPARENT'    THEN 0.5
                    WHEN k.kinship = 'GRANDCHILD'     THEN 0.5
                    WHEN k.kinship = 'AUNT_UNCLE'     THEN 0.25
                    WHEN k.kinship = 'NIECE_NEPHEW'   THEN 0.25
                    WHEN k.kinship = 'COUSIN'         THEN 0.25
                    WHEN k.kinship = 'ANCESTOR'       THEN 0.25
                    WHEN k.kinship = 'DESCENDANT'     THEN 0.25
                    WHEN k.kinship = 'RELATIVE'       THEN 0.1
                    WHEN k.kinship = 'SPOUSE'         THEN 0.5
                    WHEN k.kinship IN ('PARENT_INLAW', 'CHILD_INLAW', 'SIBLING_INLAW', 'RELATIVE_INLAW') THEN 0
                    ELSE 0
                END
            ) AS contribution,

            ROUND(COALESCE(ns.nepo_score,0))        AS person_nepo_score,
            ROUND(COALESCE(ns.parent_score,0))      AS parent_score,
            ROUND(COALESCE(ns.grandparent_score,0)) AS grandparent_score,
            ROUND(COALESCE(ns.ancestor_score,0))    AS ancestor_score,
            ROUND(COALESCE(ns.samegen_score,0))     AS samegen_score,
            ROUND(COALESCE(ns.relative_score,0))    AS relative_score
        FROM core.kinship k
        LEFT JOIN core.person p1 ON p1.person_id = k.person_id
        LEFT JOIN core.person p2 ON p2.person_id = k.related_person_id
        LEFT JOIN core.personpoint pp ON pp.person_id = k.related_person_id
        LEFT JOIN core.neposcore ns   ON ns.person_id = k.person_id
        WHERE k.person_id = :person_id
        ORDER BY contribution DESC NULLS LAST, k.kinship
        LIMIT 200
    """)

    with engine.connect() as conn:
        df = pd.read_sql(query, conn, params={"person_id": person_id})

    if df.empty:
        print(f"No kinship data found for: {identifier}")
        return None

    return df
````
